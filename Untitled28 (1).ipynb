```python

# ==========================================
# STEP 2：匯入套件及載入 AI 模型
# ==========================================

import gradio as gr
import cv2
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image, ImageEnhance, ImageFilter
import os
import tempfile

# 載入 OpenCV 人臉偵測模型
FACE_MODEL_PATH = (
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

face_cascade = cv2.CascadeClassifier(
    FACE_MODEL_PATH
)

if face_cascade.empty():
    raise RuntimeError("人臉偵測模型載入失敗！")

print("Gradio 載入成功！")
print("OpenCV 載入成功！")
print("Haar Cascade 人臉模型載入成功！")
```

```python

# ==========================================
# STEP 3：AI 影像處理核心功能
# ==========================================

def detect_faces(image):
    """偵測圖片中的人臉並回傳座標"""

    img = np.array(image.convert("RGB"))

    gray = cv2.cvtColor(
        img,
        cv2.COLOR_RGB2GRAY
    )

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    return img, faces


def draw_faces(image, faces):
    """在人臉位置繪製綠色框線"""

    result = np.array(
        image.convert("RGB")
    ).copy()

    for (x, y, w, h) in faces:

        cv2.rectangle(
            result,
            (x, y),
            (x+w, y+h),
            (0, 255, 0),
            3
        )

        cv2.putText(
            result,
            "Face",
            (x, max(y-10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )

    return Image.fromarray(result)


def apply_face_blur(image, faces):
    """人臉區域高斯模糊"""

    result = np.array(
        image.convert("RGB")
    ).copy()

    for (x, y, w, h) in faces:

        roi = result[y:y+h, x:x+w]

        if roi.size == 0:
            continue

        result[y:y+h, x:x+w] = cv2.GaussianBlur(
            roi,
            (51, 51),
            30
        )

    return Image.fromarray(result)


def apply_face_mosaic(image, faces):
    """人臉區域馬賽克"""

    result = np.array(
        image.convert("RGB")
    ).copy()

    for (x, y, w, h) in faces:

        roi = result[y:y+h, x:x+w]

        if roi.size == 0:
            continue

        small = cv2.resize(
            roi,
            (
                max(1, w//12),
                max(1, h//12)
            ),
            interpolation=cv2.INTER_LINEAR
        )

        mosaic = cv2.resize(
            small,
            (w, h),
            interpolation=cv2.INTER_NEAREST
        )

        result[y:y+h, x:x+w] = mosaic

    return Image.fromarray(result)


def apply_image_processing(
    image,
    operation,
    brightness,
    contrast
):
    """執行各種影像處理"""

    result = image.convert("RGB").copy()

    # 亮度調整
    result = ImageEnhance.Brightness(
        result
    ).enhance(brightness)

    # 對比度調整
    result = ImageEnhance.Contrast(
        result
    ).enhance(contrast)

    if operation == "原始圖片":
        pass

    elif operation == "灰階轉換":
        result = result.convert(
            "L"
        ).convert("RGB")

    elif operation == "高斯模糊":
        result = result.filter(
            ImageFilter.GaussianBlur(radius=5)
        )

    elif operation == "影像銳利化":
        result = result.filter(
            ImageFilter.SHARPEN
        )

    elif operation == "邊緣偵測":

        img = np.array(result)

        gray = cv2.cvtColor(
            img,
            cv2.COLOR_RGB2GRAY
        )

        edges = cv2.Canny(
            gray,
            100,
            200
        )

        result = Image.fromarray(
            edges
        ).convert("RGB")

    return result


def create_histogram(image):
    """建立 RGB 色彩直方圖"""

    img = np.array(
        image.convert("RGB")
    )

    fig, ax = plt.subplots(figsize=(8, 4))

    colors = ["red", "green", "blue"]

    for i, color in enumerate(colors):

        hist = cv2.calcHist(
            [img],
            [i],
            None,
            [256],
            [0, 256]
        )

        ax.plot(
            hist,
            color=color,
            label=color.upper()
        )

    ax.set_title("RGB Color Histogram")
    ax.set_xlabel("Pixel Intensity")
    ax.set_ylabel("Pixel Count")
    ax.legend()
    ax.grid(alpha=0.2)

    fig.tight_layout()

    return fig


print("所有影像處理核心功能建立完成！")
```

```python
# STEP 4：擴充功能核心（新增，不會覆蓋原本 STEP 3 函式）
import io, os, tempfile, math
from pathlib import Path

ALLOWED_FORMATS = {"PNG", "JPEG", "BMP"}

def load_original(path):
    if not path:
        raise ValueError("請先上傳 PNG、JPG/JPEG 或 BMP 圖片")
    with Image.open(path) as im:
        fmt = im.format
        if fmt not in ALLOWED_FORMATS:
            raise ValueError(f"不支援的圖片格式：{fmt}，僅接受 PNG、JPEG、BMP")
        mode = im.mode
        bands = len(im.getbands())
        width, height = im.size
        if width * height > 50_000_000:
            raise ValueError("圖片超過 5,000 萬像素，請先縮小圖片")
        rgb = im.convert("RGB")
    info = {"format": fmt, "original_mode": mode, "original_channels": bands,
            "width": width, "height": height, "size_bytes": os.path.getsize(path)}
    return rgb, info

def info_markdown(info):
    return ("### 原始影像中繼資訊\n\n"
            "| 欄位 | 值 |\n|---|---|\n"
            f"| 檔案格式 | {info['format']} |\n"
            f"| 解析度 | {info['width']} × {info['height']} px |\n"
            f"| 原始色彩模式 | {info['original_mode']} |\n"
            f"| 原始通道數 | {info['original_channels']} |\n"
            f"| 內部處理色彩模式 | RGB（3 通道） |\n"
            f"| 檔案大小 | {info['size_bytes']:,} bytes（{info['size_bytes']/1024:.2f} KiB） |")

def save_png(image, prefix="ai_vision_"):
    with tempfile.NamedTemporaryFile(prefix=prefix, suffix=".png", delete=False) as f:
        path = f.name
    image.save(path, format="PNG")
    return path

def histogram_figure(channels, title):
    fig, ax = plt.subplots(figsize=(8, 3.5))
    palette = {"R":"red", "G":"green", "B":"blue", "Y":"black", "U":"purple", "V":"orange", "Gray":"black"}
    for label, array in channels.items():
        counts = np.bincount(np.asarray(array, dtype=np.uint8).ravel(), minlength=256)
        ax.plot(np.arange(256), counts, color=palette.get(label, "gray"), label=label)
    ax.set_xlim(0,255)
    ax.set_xlabel("Pixel Intensity (0–255)")
    ax.set_ylabel("Pixel Count")
    ax.set_title(title)
    ax.grid(alpha=.2)
    ax.legend()
    fig.tight_layout()
    return fig

def color_space_view(path, mode):
    original, info = load_original(path)
    rgb = np.asarray(original)
    if mode == "Original (RGB)":
        return original, histogram_figure({"R":rgb[:,:,0], "G":rgb[:,:,1], "B":rgb[:,:,2]}, "Original RGB Histogram"), info_markdown(info)
    if mode in ("R", "G", "B"):
        i = {"R":0, "G":1, "B":2}[mode]
        # 單通道灰階顯示，避免將其他通道填 0 誤解為原始強度
        chan = rgb[:,:,i]
        return Image.fromarray(chan), histogram_figure({mode:chan}, f"RGB {mode} Channel Histogram"), info_markdown(info)
    yuv = cv2.cvtColor(rgb, cv2.COLOR_RGB2YUV)
    i = {"Y":0, "U":1, "V":2}[mode]
    chan = yuv[:,:,i]
    return Image.fromarray(chan), histogram_figure({mode:chan}, f"YUV {mode} Channel Histogram"), info_markdown(info) + "\n\nYUV 分量以 OpenCV 8-bit 編碼值顯示；U/V 的 128 附近代表中性色度。"

KERNEL_PRESETS = {
    "單位矩陣": "0,0,0\n0,1,0\n0,0,0",
    "均值平滑": "1,1,1\n1,1,1\n1,1,1",
    "高斯平滑": "1,2,1\n2,4,2\n1,2,1",
    "銳化": "0,-1,0\n-1,5,-1\n0,-1,0",
    "浮雕": "-2,-1,0\n-1,1,1\n0,1,2",
}

def parse_kernel(text):
    if not text or not text.strip():
        raise ValueError("卷積核不能為空")
    rows = []
    for line in text.strip().splitlines():
        if not line.strip():
            continue
        vals = [float(x.strip()) for x in line.replace("，", ",").replace(" ", ",").split(",") if x.strip()]
        rows.append(vals)
    if not rows or len({len(r) for r in rows}) != 1:
        raise ValueError("每一列的數值數量必須相同")
    h, w = len(rows), len(rows[0])
    if h != w or h % 2 != 1 or h < 3 or h > 15:
        raise ValueError("請輸入 3×3、5×5 等奇數正方矩陣（最大 15×15）")
    kernel = np.array(rows, dtype=np.float64)
    if not np.isfinite(kernel).all():
        raise ValueError("矩陣不得含 NaN 或 Infinity")
    if np.max(np.abs(kernel)) > 1e6:
        raise ValueError("權重絕對值不得超過 1,000,000")
    return kernel

def psnr_rgb(a, b):
    aa = np.asarray(a.convert("RGB"), dtype=np.float64)
    bb = np.asarray(b.convert("RGB"), dtype=np.float64)
    mse = float(np.mean((aa-bb)**2))
    return mse, math.inf if mse == 0 else 10 * math.log10(255**2/mse)

def kernel_process(path, matrix_text, normalize, border_name):
    original, info = load_original(path)
    k = parse_kernel(matrix_text)
    raw_sum = float(k.sum())
    if normalize:
        if abs(raw_sum) < 1e-12:
            raise ValueError("卷積核總和為 0，不能以總和正規化；請取消正規化")
        k = k / raw_sum
    borders = {"反射 (Reflect101)":cv2.BORDER_REFLECT_101,
               "複製邊界 (Replicate)":cv2.BORDER_REPLICATE,
               "固定零值 (Constant)":cv2.BORDER_CONSTANT}
    rgb = np.asarray(original, dtype=np.float32)
    filtered = cv2.filter2D(rgb, ddepth=-1, kernel=k, borderType=borders[border_name])
    filtered = Image.fromarray(np.clip(np.rint(filtered),0,255).astype(np.uint8))
    mse, psnr = psnr_rgb(original, filtered)
    result = (f"### 卷積結果\n\n卷積核尺寸：{k.shape[0]} × {k.shape[1]}；"
              f"原始權重總和：{raw_sum:.6g}；正規化：{'是' if normalize else '否'}；"
              f"邊界處理：{border_name}。\n\n"
              f"**MSE = {mse:.6f}**；**PSNR = {'∞（影像完全相同）' if math.isinf(psnr) else f'{psnr:.3f} dB'}**。\n\n"
              "PSNR 為相對原圖的像素差異指標，並不代表視覺品質一定較佳。")
    return original, filtered, result, save_png(filtered, "kernel_")

def face_detect_with_weights(original):
    rgb = np.asarray(original.convert("RGB"))
    gray = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY)
    faces, reject_levels, level_weights = face_cascade.detectMultiScale3(
        gray, scaleFactor=1.1, minNeighbors=5, minSize=(30,30), outputRejectLevels=True)
    output = rgb.copy()
    rows = []
    for i, (x,y,w,h) in enumerate(faces):
        x,y,w,h = map(int,(x,y,w,h))
        cv2.rectangle(output,(x,y),(x+w,y+h),(0,255,0),2)
        weight = float(level_weights[i]) if i < len(level_weights) else float('nan')
        cv2.putText(output,f"Face {i+1}",(x,max(y-8,15)),cv2.FONT_HERSHEY_SIMPLEX,.6,(0,255,0),2)
        rows.append(f"| {i+1} | ({x}, {y}, {w}, {h}) | {weight:.3f} |")
    table = "\n".join(rows) if rows else "| — | 無偵測結果 | — |"
    report = (f"### Haar-Cascade 人臉偵測\n\n**偵測人臉數：{len(faces)}**\n\n"
              "| 人臉 | 邊界框 (x, y, w, h) | levelWeight |\n|---|---|---|\n"+table+
              "\n\n**注意：** `levelWeight` 是 OpenCV Cascade 偵測器輸出的分類權重，"
              "不是校準後的機率或 0–100% Confidence Score，不能當作百分比。"
              "此版本未實作可校準的機率信心度；若驗收要求機率值，需另接具備明確分數定義的模型。")
    return Image.fromarray(output), report, save_png(Image.fromarray(output), "faces_")

print("STEP 4 擴充功能載入完成")
```

```python
# STEP 4B：新版人臉偵測：SSD + Confidence + NMS
# confidence 是模型輸出的偵測分數，不代表校準後的真實正確機率。

def dnn_detect_faces(original, threshold=0.60, nms_iou=0.35):
    """融合 SSD 與 YuNet，回傳 (x,y,w,h,score)；score 為各模型原始偵測分數。"""
    rgb = np.asarray(original.convert('RGB'))
    h, w = rgb.shape[:2]
    bgr = cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR)
    candidates = []

    # SSD：原有偵測器，保留對正面人臉的支援
    blob = cv2.dnn.blobFromImage(cv2.resize(bgr, (300, 300)), 1.0,
                                 (300, 300), (104.0, 177.0, 123.0))
    face_net.setInput(blob)
    predictions = face_net.forward()
    for item in predictions[0, 0]:
        score = float(item[2])
        if score < threshold:
            continue
        x1, y1, x2, y2 = (item[3:7] * np.array([w, h, w, h])).astype(int)
        x1, y1 = max(0, min(w - 1, x1)), max(0, min(h - 1, y1))
        x2, y2 = max(0, min(w, x2)), max(0, min(h, y2))
        if x2 > x1 and y2 > y1:
            candidates.append((x1, y1, x2-x1, y2-y1, score, 'SSD'))

    # YuNet：不同姿態的補充偵測；保留原始圖片解析度
    yunet_detector.setInputSize((w, h))
    yunet_detector.setScoreThreshold(float(threshold))
    _, results = yunet_detector.detect(bgr)
    if results is not None:
        for row in results:
            x, y, bw, bh = map(float, row[:4])
            score = float(row[-1])
            x1, y1 = max(0, int(round(x))), max(0, int(round(y)))
            x2, y2 = min(w, int(round(x + bw))), min(h, int(round(y + bh)))
            if x2 > x1 and y2 > y1:
                candidates.append((x1, y1, x2-x1, y2-y1, score, 'YuNet'))

    if not candidates:
        return []

    # 跨模型去重：使用 IoU，而不是直接比較不同模型的分數大小
    def iou(a, b):
        ax1, ay1, aw, ah = a[:4]
        bx1, by1, bw, bh = b[:4]
        xx1, yy1 = max(ax1, bx1), max(ay1, by1)
        xx2, yy2 = min(ax1+aw, bx1+bw), min(ay1+ah, by1+bh)
        inter = max(0, xx2-xx1) * max(0, yy2-yy1)
        union = aw*ah + bw*bh - inter
        return inter / union if union else 0.0

    # YuNet 優先：解析度較高，避免被 SSD 的大框覆蓋
    ordered = sorted(candidates, key=lambda f: (f[5] == 'YuNet', f[4]), reverse=True)
    kept = []
    for candidate in ordered:
        if all(iou(candidate, prior) < nms_iou for prior in kept):
            kept.append(candidate)
    return [f[:5] for f in sorted(kept, key=lambda f: (f[1], f[0]))]

def face_detect_dnn(original, threshold=0.60):
    faces = dnn_detect_faces(original, float(threshold))
    output = np.asarray(original.convert('RGB')).copy()
    rows = []
    for i, (x, y, w, h, confidence) in enumerate(faces, start=1):
        cv2.rectangle(output, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(output, f'Face {i}: {confidence:.2f}',
                    (x, max(18, y-7)), cv2.FONT_HERSHEY_SIMPLEX,
                    0.5, (0, 255, 0), 2)
        rows.append(f'| {i} | ({x}, {y}, {w}, {h}) | {confidence:.3f} |')
    report = (f'### SSD + YuNet 雙模型人臉偵測\n\n**偵測人臉數：{len(faces)}**；'
              f'偵測門檻：{threshold:.2f}\n\n'
              '| 編號 | Bounding Box (x, y, w, h) | Confidence Score |\n'
              '|---|---|---|\n' + ('\n'.join(rows) if rows else '| — | 無偵測結果 | — |') +
              '\n\n註：Confidence 為模型輸出的偵測分數，並非經機率校準的準確率。')
    result = Image.fromarray(output)
    return result, report, save_png(result, 'dnn_faces_')

def face_detect_path(path, threshold):
    if not path:
        raise gr.Error('請先上傳圖片')
    original, _ = load_original(path)
    return face_detect_dnn(original, threshold)
```
